```python
import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Correct Yahoo ticker for NIFTY 50
ticker = 'HDFCBANK.NS'

# Download daily NIFTY data from 2018 to 2024
data = yf.download(ticker, start="2008-01-01", end="2024-01-01", interval='1d')

# Save to CSV
data.to_csv('nifty.csv')

print("Data downloaded and saved as nifty.csv")
print(data.head())
```
