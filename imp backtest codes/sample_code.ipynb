```python
import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
# import 
# Correct Yahoo ticker for NIFTY 50
import pandas as pd

# Go one folder up, then into stock_data folder
df = pd.read_csv('../stock_data/ind_nifty50list.csv', index_col=0, parse_dates=True)
symbol=df['Symbol']
symbol.tolist()
for i in symbol:
    ticker = i+'.NS'
    # Download daily NIFTY data from 2018 to 2024
    data = yf.download(ticker, start="2020-09-01", end="2025-09-01", interval='1d')
    data.columns = data.columns.droplevel(1)
    # Save to CSV
    filename=i+'.csv'
    data.to_csv('../stock_data/'+filename)
    print(f"Data for {i} downloaded and saved as {filename}")
    print(data.head(1))
```
